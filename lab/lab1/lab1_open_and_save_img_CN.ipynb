```python
# firstly, you must install some required packages and then import opencv-python numpy matplotlib.
import numpy as np
import cv2
from matplotlib import pyplot as plt
from PIL import Image
```

## 读取并打开图像

`cv.imread` -- 使用 **OpenCV** 从文件中加载图像。

`imread` 函数会从指定文件加载图像并返回 **OpenCV 矩阵**。如果图像无法读取（例如文件缺失、权限不足或格式无效/不受支持），函数会返回空矩阵。

当前支持的文件格式包括：

- Windows 位图 - \*.bmp, \*.dib（始终支持）

- GIF 文件 - \*.gif（始终支持）

- JPEG 文件 - \*.jpeg, \*.jpg, \*.jpe（见注释）

- JPEG 2000 文件 - \*.jp2（见注释）

- PNG 文件 - \*.png（见注释）

- WebP - \*.webp（见注释）

- AVIF - \*.avif（见注释）

- 可移植图像格式 - \*.pbm, \*.pgm, \*.ppm, \*.pxm, \*.pnm（始终支持）

- PFM 文件 - \*.pfm（见注释）

- Sun raster - \*.sr, \*.ras（始终支持）

- TIFF 文件 - \*.tiff, \*.tif（见注释）

- OpenEXR 图像文件 - \*.exr（见注释）

- Radiance HDR - \*.hdr, \*.pic（始终支持）


```python
# cv.imread
img =cv2.imread('lenna.jpg')   # Read image in BGR mode
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)  # Convert BGR into RGB
print(img.shape)# here use image shape
print(img[:,:,0])

plt.imshow(img)
plt.xticks([]),plt.yticks([])  # to hide tick values on X and Y axis
plt.show()
```

`PIL.Image.open` -- 使用 **PIL** 打开并识别给定图像文件。

返回一个 `Image` 对象。你可以使用 `Image` 的方法方便地对图像进行处理，处理后仍会返回 `Image` 对象。

```python
#PIL.Image.open
img_pil = Image.open('lenna.jpg')

print(img_pil, img_pil.size)# here use Image.size
img_pil.show()# There will be a open-window to show this image
```

## 相互转换

```python
# array(cv.imread)-->Image(PIL.Image.open)
img_PIL = Image.fromarray(img)

print(img_PIL, img_PIL.size)
img_PIL.show()
```

```python
# Image(PIL.Image.open)-->array(cv.imread)
img_cv2 = np.array(img_pil)

print(img_cv2.shape)

plt.imshow(img_cv2)
plt.xticks([]),plt.yticks([])  # to hide tick values on X and Y axis
plt.show()
```

## 图像写入与保存

```python
# cv.imread-->cv.imwrite
img =cv2.imread('lenna.jpg')   # Read image in BGR mode
img_gray= cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)  # Convert BGR into GRAY

cv2.imwrite('lenna_gry.jpg',img_gray)
```

```python
# PIL.Image.open-->PIL.Image.save
img_pil = Image.open('lenna.jpg')
img_pil=img_pil.convert("L")

img_pil.save("lenna_gry_pil.jpg")
```
